Seaborn gallery is an incredibly useful resource in how to make fairly in-depth graphs. Unfortunately it can be very difficult at times to figure out exactly what the example code is actually doing with each variable and when the best time to use each graph. The following will list all of the Seaborn gallery plots, when to best use them, and what each variable in the graph actually is. 

# Implot (Ansecombe's Quartet)

```python
import seaborn as sns

# Import data
dataframe = sns.load_dataset("penguins")

#Specific graph set up for a clean, blank white background graph, with minimal grid lines
#Other options include "darkgrid", "whitegrid", "white", "dark"
sns.set_theme(style="ticks")

plot = sns.lmplot(
    data=dataframe,
    x="bill_length_mm", #Numeric X-variable
    y="bill_depth_mm", #Numeric Y-variable
    col="species", #Categorical variable that is different between each graph
    hue="species", #Categorical color variable that is different between each graph
    
    col_wrap=2, #How many mini-graphs are allowed on one row
    palette="colorblind", #Overall color scheme that is not impacted by the data
    ci=None, #Option to have a shaded confidence interval surrounding the regression lines on the plots
    height=4, #Height of the plots (in inches)
    scatter_kws={"s": 50, "alpha": 1} #Looks at the dots of the graphs, s controls size, alpha controls opacity (0 is see through, 1 is 
)
#Change labels with the below commands
plot.set_axis_labels("Bill Length (mm)", "Body Mass (g)")  
plot.fig.suptitle("Penguin Body Mass vs Bill Length by Species", y=1.02) #y is used to determine placement of text, it determines distance from the label to the top of a plot
```

<b> Question this visual answers: Do different categories create patterns between the same numeric variables? </b>
The best use of this data visualization would be for data that has a categorical variable with which one is examining the differences in numeric variables. Meaning that if you have a category that can change, like species in the penguin data set, and numeric variables that you are examining based on that categorical change, like length and depth of a penguin's bill based on their species, this is a perfect graph to view all that data right next to each other.

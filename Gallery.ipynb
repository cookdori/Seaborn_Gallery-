# Implot (Ansecombe's Quartet)

```python
import seaborn as sns

# Import data
dataframe = sns.load_dataset("penguins")

#Specific graph set up for a clean, blank white background graph, with minimal grid lines
#Other options include "darkgrid", "whitegrid", "white", "dark"
sns.set_theme(style="ticks")

plot = sns.lmplot(
    data=dataframe,
    x="bill_length_mm", #Numeric X-variable
    y="bill_depth_mm", #Numeric Y-variable
    col="species", #Categorical variable that is different between each graph
    hue="species", #Categorical color variable that is different between each graph
    
    col_wrap=2, #How many mini-graphs are allowed on one row
    palette="colorblind", #Overall color scheme that is not impacted by the data
    ci=None, #Option to have a shaded confidence interval surrounding the regression lines on the plots
    height=4, #Height of the plots (in inches)
    scatter_kws={"s": 50, "alpha": 1} #Looks at the dots of the graphs, s controls size, alpha controls opacity (0 is see through, 1 is 
)
#Change labels with the below commands
plot.set_axis_labels("Bill Length (mm)", "Body Mass (g)")  
plot.fig.suptitle("Penguin Body Mass vs Bill Length by Species", y=1.02) #y is used to determine placement of text, it determines distance from the label to the top of a plot
```

<b> Question this visual answers: Do different categories create patterns between the same numeric variables? </b> <br>
The best use of this data visualization would be for data that has a categorical variable with which one is examining the differences in numeric variables. Meaning that if you have a category that can change, like species in the penguin data set, and numeric variables that you are examining based on that categorical change, like length and depth of a penguin's bill based on their species, this is a perfect graph to view all that data right next to each other.

***

# Scatterplot (Scatterplot with multiple schematics)

```python
import seaborn as sns
import matplotlib.pyplot as plt

#Import data
dataframe = sns.load_dataset("penguins")

#Specific graph set up for a clean, blank white background graph, with grid lines
#Other options include "darkgrid", "ticks", "white", "dark"
sns.set_theme(style="whitegrid")

#Create the backdrop of the plot itself, f is the whole picture, ax is where the plot goes
f, ax = plt.subplots(figsize=(6.5, 6.5)) #figsize controls the dimensions of the plot in inches

sns.despine(f, left=True, bottom=True) #Removes the edges from the plot (the x and y axis lines), but not the grid

sns.scatterplot(
    data = dataframe,
    x = "flipper_length_mm", #Numeric X-variable
    y="body_mass_g", #Numeric Y-Variable
    hue="species", #Categorical Color variable
    size="bill_depth_mm", #Numeric or categorical size variable (if using numeric, values should be relatively evenly spaced)
    palette="colorblind", #Overall color scheme, not impacted by data
    sizes=(20, 200), #Controls min and max sizes the dots can get
    linewidth=0, #Data points can have boarders, 0 is no boarders, 1 is yes boarders 
    ax=ax #Actually sets up the graph in the space that you created for it earlier 
)

#Changing Labels
ax.set_title("Scatterplot")
ax.set_xlabel("X_label")
ax.set_ylabel("Y_label")
legend = ax.legend()
legend.set_title("Penguin Features")
```

<b> Question this visual answers: How do multiple numeric and categorical variables work together in my data? </b> <br>
The best use of this data visualization would be more geared toward data exploration and seeing overall trends in the data set. There are, in this case, four variables represented, which is many more than most data visualization types allow or look good. This plot should be used with at least one categorical variable. The shape of each data point could be another categorical variable if necessary. However, this plot should not be used when there are numerous categories in one variable, as all the different colors/shapes/data point sizes can get confusing.

# Lineplot (time series plot with error bands)

```python
import seaborn as sns

#Specific graph set up for a clean, blank dark background graph, with grid lines
#Other options include "whitegrid", "ticks", "white", "dark"
sns.set_theme(style="darkgrid")

#Import Data
Dataframe = sns.load_dataset("fmri")
#SNS fmri data is looking at blood-oxygen-level-dependent signal changes overtime in the brain

sns.lineplot(x = "timepoint", #Numeric X Variable (continuous)
             y = "signal", #Numeric Y Variable
             hue = "region", #Categorical Color Variable
             style = "event", #Categorical Line style Variable
             data = Dataframe)

#Changing labels
plt.title("Progression Graph")
plt.xlabel("Time")
plt.ylabel("Y label (Signal)")
```

<b> Question this visual answers: How do different variables change over a continuous variable? </b> <br>
The best use of this data visualization would be if you have a numeric time variable or another way of tracking progress (different stages, etc.) along with another numeric variable (y_axis). This graph can be created without the line and color changes  Unfortunately the regular penguins data set that I like does not have such a variable and therefore does not lend itself to this visualization.

# Displot (Facetting histograms with subsets of data)

```python
import seaborn as sns

#Import data
dataframe = sns.load_dataset("penguins")

#Specific graph set up for a clean, blank dark background graph, with grid lines
#Other options include "whitegrid", "ticks", "white", "dark"
sns.set_theme(style="darkgrid")

plot = sns.displot(
    data = dataframe,
    x = "flipper_length_mm", #Numeric X Variable (could be categorical)
    col = "species", #Categorical Variable that changes between columns
    row = "sex", #Categorical Variable that changes between rows
    binwidth = 4, #Size of the bars in each graph
    height = 3, # Height of each graph (Subsequently width too as they are squares)
    facet_kws = dict(margin_titles=True), #Special Labeling call, facet_kws is a parameter for extra settings, margin_titles control if there are margin titles (sex and species in this case)
)

#Changing Labels 
plot.set_axis_labels("Flipper length (mm)", "Count")  
```

<b> Question this visual answers: How does the distribution of a numeric or categorical variable change across different categorical variables? </b> <br>
The best use of this data visualization would be if you are looking at only one variable and are more interested in its abundance than its actual value, across two different categorical variables. It's very good when you're trying to show differences between those two categorical variables as well as showing a lot of information about the data set at one time. It also pulls in the benefits of looking at a bar graph, those being looking at distribution, skew, and other basic statistic patterns.

# Relplot (Lineplots on multiple facets)

```python
import seaborn as sns

#import data, penguins doesn't have a time series and therefore doesn't well fit this visualization
dataframe = sns.load_dataset("dots")

#Specific graph set up for a clean, blank white background graph, without grid lines
#Other options include "whitegrid", "darkgrid", "white", "dark"
sns.set_theme(style="ticks")

plot = sns.relplot(
    data = dataframe,
    x = "time", #Numeric X Variable
    y = "firing_rate", #Numeric Y Variable
    hue = "coherence", #Categorical or Numeric color variable (If using numeric data, binning may be necessary)
    size = "choice", #Categorical Variable controlling the thickness of the lines
    col = "align", #Categorical Variable separating subplots
    kind = "line", #Makes lines plotted instead of a scatterplot
    size_order = ["T1", "T2"], #Assigns values for the earlier "size = "choice"" line
    palette = "colorblind", #Overall color scheme, is not impacted by data
    height = 5, #Height of each subplot (in inches)
    aspect = .75, #Width of each subplot relative to height (width = height * aspect in inches)
    facet_kws = dict(sharex=False), #Extra setting parameter that makes each subplot be able to have its own x axis
)

#Changing axis labels
plot.set_axis_labels("Time", "Firing Rate")
```

<b> Question this visual answers: How does a numeric variable change over a continuous variable with respect to different groups and conditions? </b> <br>
The best use of this data visualization would be if you have a data set with a continuous variable, like time and want to look at the changes in multiple categorical and numeric variables across time or whatever continuous variable you have. It provides a broad look at the data from a macroscopic perspective to identify patterns within subgroups of variables.

# Catplot (Grouped Barplots)

```python
import seaborn as sns

#import data
dataframe = sns.load_dataset("penguins")

#Specific graph set up for a clean, blank white background graph, with grid lines
#Other options include "ticks", "darkgrid", "white", "dark"
sns.set_theme(style="whitegrid")

plot = sns.catplot(
    data = dataframe,
    kind = "bar", #Specifies the exact kind of graph, a bar chart in this case
    x = "species", #Categorical x variable
    y = "body_mass_g", #Numeric y variable (can also be count)
    hue = "sex", #Categorical color variable
    errorbar = "sd", #Error bars are the standard deviation of the body mass of each bird sex and species
    palette = "colorblind", #Overall colorscheme, not impacted by the data
    alpha=.6, #Opacity, 0 is very see through, 1 is not see through at all. Helpful when data is clustered and you want to see multiple layers of it
    height=6 #Height of graph in inches
)

#Changing Labels
plot.despine(left=True) #Takes out the y axis line
plot.set_axis_labels("", "Body mass (g)")
plot.legend.set_title("")
```

<b> Question this visual answers: How does the distribution of a variable change across other different variables? </b> <br>
The best use of this data visualization would be if you want to get a more statistical look at variables in your data set. Bar plots themselves are great looks at distributions for skew, approximate mean and median location, and standard deviation values. This specific plot is a good subdivider of these distributions across other variables and give a wholistic first look at a data set.
